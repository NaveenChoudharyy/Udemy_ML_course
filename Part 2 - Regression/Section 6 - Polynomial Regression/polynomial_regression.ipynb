# Polynomial Regression

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1] # We dont want position column coz both Level and Positions have all uniques and it won't make sense to use both
y = dataset.iloc[:, -1]
```

```python
X.head()
```

#### Not splitting into train and split coz we need max data for poly regression

## Training the Linear Regression model on the whole dataset

```python
from sklearn.linear_model import LinearRegression
lin_regressor = LinearRegression()
lin_regressor.fit(X, y)
```

## Training the Polynomial Regression model on the whole dataset

```python
from sklearn.preprocessing import PolynomialFeatures
new_poly_fetures = PolynomialFeatures(degree=2)
poly_fetures = new_poly_fetures.fit_transform(X)
```

```python
poly_fetures
```

```python
poly_regressor = LinearRegression()
poly_regressor.fit(poly_fetures, y)
```

## Visualising the Linear Regression results

```python
y_pred_lin = lin_regressor.predict(X)
```

```python
plt.scatter(X, y, color="red")
plt.plot(X, y_pred_lin, color="blue")
plt.title("Truth or Bulff (Linear Regression)")
plt.xlabel("Lavel")
plt.ylabel("Salary")
plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

## Visualising the Polynomial Regression results

```python
y_pred_poly = poly_regressor.predict(poly_fetures)
```

```python
plt.scatter(X, y, color="red")
plt.plot(X, y_pred_poly, color="blue")
plt.title("Truth or Bulff (Polynomial Regression)")
plt.xlabel("Lavel")
plt.ylabel("Salary")
plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

## Predicting a new result with Linear Regression

```python
lin_regressor.predict([[6]])
```

## Predicting a new result with Polynomial Regression

```python
poly_regressor.predict(new_poly_fetures.fit_transform([[6.3]]))
```
