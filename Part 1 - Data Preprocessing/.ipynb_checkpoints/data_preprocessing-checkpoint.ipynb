# Data Preprocessing Tools

## Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Importing the dataset

```python
dataset = pd.read_csv(r"C:\Users\Naveen\Desktop\GitHub\Udemy_ML_course\Part 1 - Data Preprocessing\Data.csv")
dataset.head()
```

```python
X = dataset.loc[:, ["Country", "Age", "Salary"]]
y = dataset.loc[:, ["Purchased"]]
```

```python
display(X[:5])
print("--------------------------------------")
display(y[:5])
```

## Taking care of missing data

```python
def Fillna_ContVars(df):
    X = df.select_dtypes(["int", "float"])
    
    for i in X.columns:
        X[i] = X[i].fillna(X[i].median())

    df[X.columns] = X
    return df
```

```python
Fillna_ContVars(X)
```

## Encoding categorical data

```python
# One hot-encoding

X = pd.get_dummies(X, drop_first=True).astype(float)
```

### Encoding the Independent Variable

```python
X
```

### Encoding the Dependent Variable

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
y = le.fit_transform(y)
y
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split

train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.3)
```

```python
display(train_X)
print("-----------------------")
display(train_y)
print("-----------------------")
display(test_X)
print("-----------------------")
display(test_y)
print("-----------------------")
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
std_sc = StandardScaler()

# -------------------------------We don't want to use StandardScaler on dummies or encoded values 
std = StandardScaler()

# For train data
std.fit(train_X.iloc[:, :2])
train_X.iloc[:, :2] = std.transform(train_X.iloc[:, :2])

# For test

test_X.iloc[:, :2] = std.transform(test_X.iloc[:, :2])
```
